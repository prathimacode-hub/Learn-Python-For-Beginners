### Sets & Frozenset

When working with data in Python, you might come across situations where you need to use sets or frozensets. Sets and frozensets are two data types that are commonly used in Python to store collections of unique elements. While both are similar in many ways, there are some important differences between them. In this blog post, we will explore the differences between sets and frozensets, and when you should use one over the other.

Set and frozenset are two built-in collection data types in Python that are used to store a collection of unique elements. While set is mutable, meaning that we can add, remove, or change elements in a set, frozenset is immutable and cannot be modified after creation.

#### Differences between set and frozenset

Mutability: As mentioned earlier, set is mutable while frozenset is immutable. This means that we can add or remove elements from a set, but not from a frozenset.

Memory usage: Since set is mutable, it requires more memory to store the elements as compared to frozenset, which is immutable and requires less memory.

Hashability: A set can only contain hashable objects, i.e., objects that have a hash value that remains constant throughout their lifetime. On the other hand, frozenset can contain any immutable object, including other frozensets.

Iteration: Both set and frozenset support iteration, but since frozenset is immutable, it can be iterated more efficiently than a set.

#### Use cases of set and frozenset

1. Removing duplicates: One of the most common use cases of set is to remove duplicates from a list or any other sequence. Since sets only contain unique elements, we can convert the sequence to a set, which automatically removes the duplicates.

```python
my_list = [1, 2, 3, 2, 1, 4, 5, 4]
my_set = set(my_list)
print(my_set)  # Output: {1, 2, 3, 4, 5}
```

2. Membership testing: Both set and frozenset are efficient for membership testing. We can use the in operator to check if an element is present in a set or a frozenset.

```python
my_set = {1, 2, 3, 4, 5}
if 3 in my_set:
    print("Element found")
else:
    print("Element not found")
```

3. Storing hashable elements: Since sets only contain hashable elements, they are useful when we need to store a collection of unique hashable elements.

```python
my_set = {'apple', 'banana', 'cherry'}
```

4. Hashing keys in a dictionary: Since frozenset is immutable, it can be used as a key in a dictionary, while set cannot be used as a key since it is mutable.

```python
my_dict = {frozenset({1, 2}): 'value'}
print(my_dict)  # Output: {frozenset({1, 2}): 'value'}
```

5. Memory usage: In terms of memory usage, frozenset takes up less memory than a regular set. This is because a set is mutable, meaning that its elements can be added, removed, or changed after it is created. As a result, a set needs to allocate extra memory to allow for these changes. On the other hand, a frozenset is immutable, so it doesn't need this extra memory allocation, making it more memory-efficient.

To demonstrate this, you can use the sys.getsizeof() function in Python to check the memory usage of objects.

```python
import sys

s = {1, 2, 3, 4, 5}
fs = frozenset(s)

print(sys.getsizeof(s))  # output: 736
print(sys.getsizeof(fs)) # output: 224
```

To summarize, set and frozenset are both important data types in Python that offer unique features for handling collections of unique elements. set is mutable and can be modified, while frozenset is immutable and cannot be modified after creation. set offers faster operations for modifications, while frozenset offers better memory efficiency for read-only collections.

When choosing between set and frozenset, the decision should be based on the specific use case and requirements of the application. If you need a collection that can be modified or updated over time, then set is the better choice. On the other hand, if you have a read-only collection that will not change, then frozenset provides better memory efficiency and can be used as a hashable key for dictionaries and other data structures.

### LinkedList

A linked list in Python is a linear data structure composed of a sequence of nodes, where each node contains data and a reference (or pointer) to the next node in the sequence. Unlike Python's built-in list type, which functions as a dynamic array and stores elements contiguously in memory, linked lists store nodes at potentially non-contiguous memory locations. The order is maintained through the explicit links between nodes.
Key components of a linked list in Python:
    
Node Class:
Each node is typically represented as an object of a Node class.

A Node object contains two main attributes:
    
data: The value or information stored within that node.
    
next: A reference to the next node in the linked list. This reference is None for the last node in the list. 

```python
    class Node:
        def __init__(self, data):
            self.data = data
            self.next = None
```

#### LinkedList Class:

This class manages the entire linked list.

It typically contains a head attribute, which is a reference to the first node in the list. If the list is empty, head is None.


Methods are added to this class for various operations, such as:


Insertion: Adding new nodes (e.g., at the beginning, end, or a specific position).

Deletion: Removing nodes from the list.

Traversal: Iterating through the nodes of the list to access their data.

Searching: Finding a specific node based on its data.

```python
    class LinkedList:
        def __init__(self):
            self.head = None

        def insert_at_beginning(self, data):
            new_node = Node(data)
            new_node.next = self.head
            self.head = new_node

        def display(self):
            current = self.head
            while current:
                print(current.data, end=" -> ")
                current = current.next
            print("None")
```

#### Advantages of Linked Lists:
    
Dynamic Size:
Linked lists can easily grow or shrink in size as needed, without requiring pre-allocation of memory.

Efficient Insertions and Deletions:
Adding or removing elements at arbitrary positions can be more efficient than with arrays, as it primarily involves updating pointers rather than shifting elements.

#### Disadvantages of Linked Lists:
    
No Random Access:
Accessing an element by its index requires traversing the list from the beginning, leading to O(n) time complexity for access.

Increased Memory Overhead:
Each node requires extra memory to store the next reference (and potentially a prev reference in a doubly linked list).

### LinkedList Traversal

Linked list traversal in Python involves iterating through each node of the linked list, starting from the head node, until the end of the list is reached. 

#### Steps for Linked List Traversal:

- Initialize a pointer:
Create a temporary pointer, often named current_node or temp, and set it to the head of the linked list. The head is the first node in the list.

- Iterate through the list:
Use a while loop that continues as long as current_node is not None. The next attribute of the last node in a linked list typically points to None, signifying the end of the list.

- Process the current node:
Inside the loop, perform the desired operation on the current_node. This could involve printing its data, checking for a specific value, or modifying its contents.

- Move to the next node:
Update current_node to point to the next node in the sequence by setting current_node = current_node.next. This moves the pointer forward to the subsequent node in the linked list. 

#### Example Implementation:

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last_node = self.head
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node

    def traverse_and_print(self):
        current_node = self.head
        while current_node:
            print(current_node.data, end=" -> ")
            current_node = current_node.next
        print("None")

# Usage example
my_list = LinkedList()
my_list.append(10)
my_list.append(20)
my_list.append(30)
my_list.traverse_and_print() # Output: 10 -> 20 -> 30 -> None
```
